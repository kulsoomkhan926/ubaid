## Online Order System

    Ask:
    
    Order amount
    
    User type (regular / premium)

    Premium → 20% discount

```python
order_amount = float(input("Enter order amount: "))
user_type = input("Enter user type (regular/premium): ").strip().lower()

if order_amount >= 1000:
    if user_type == "premium":
        discount = order_amount * 0.20
        final_amount = order_amount - discount
        print(f"20% discount applied! Discount: {discount}, Final Amount: {final_amount}")
    elif user_type == "regular":
        discount = order_amount * 0.10
        final_amount = order_amount - discount
        print(f"10% discount applied! Discount: {discount}, Final Amount: {final_amount}")
    else:
        print("Invalid user type entered.")
else:
    print("No discount available")
```

## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed = float(input("Enter vehicle speed (km/h): "))
is_emergency = input("Is it an emergency vehicle? (yes/no): ").strip().lower()
is_rainy = input("Is the weather rainy? (yes/no): ").strip().lower()

if is_emergency == "yes":
    print("No fine (Emergency vehicle)")
else:
    if speed <= 60:
        print("No fine")
    elif 61 <= speed <= 80:
        base_fine = 500
        fine = base_fine * 1.5 if is_rainy == "yes" else base_fine
        print(f"Fine: Rs. {int(fine)}")
    elif 81 <= speed <= 100:
        base_fine = 2000
        fine = base_fine * 1.5 if is_rainy == "yes" else base_fine
        print(f"Fine: Rs. {int(fine)}")
    else:
        print("License suspended")
```

## A company filters candidates:
* Experience ≥ 5 years AND skill score ≥ 80 → shortlisted
* Experience 2–4 years AND skill score ≥ 70 → interview
* Freshers → only if test score ≥ 85
* If criminal record = True → reject immediately
* If multiple job switches (>3 in 2 years) → flag as unstable

```python
criminal_record = input("Criminal record? (true/false): ").strip().lower() == "true"

if criminal_record:
    print("Rejected immediately (Criminal record)")
else:
    job_switches = int(input("Number of job switches in the last 2 years: "))
    
    if job_switches > 3:
        print("Flagged as unstable")
    
    experience = float(input("Enter years of experience: "))
    
    if experience >= 5:
        skill_score = float(input("Enter skill score: "))
        if skill_score >= 80:
            print("Status: Shortlisted")
        else:
            print("Status: Not shortlisted")
    elif 2 <= experience <= 4:
        skill_score = float(input("Enter skill score: "))
        if skill_score >= 70:
            print("Status: Called for Interview")
        else:
            print("Status: Not shortlisted for interview")
    elif experience < 2:
        test_score = float(input("Enter test score: "))
        if test_score >= 85:
            print("Status: Shortlisted (Fresher)")
        else:
            print("Status: Rejected (Low test score)")
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
passport_valid = input("Is the passport valid? (yes/no): ").strip().lower()
visa_status = input("Visa status (valid/invalid): ").strip().lower()
criminal_check = input("Criminal check status (clear/flagged): ").strip().lower()

if passport_valid != "yes":
    print("Status: Rejected (Invalid passport)")
elif visa_status != "valid":
    print("Status: Rejected (Invalid visa)")
elif criminal_check != "clear":
    print("Status: Security Hold (Criminal check flagged)")
else:
    baggage_weight = float(input("Enter baggage weight (in kg): "))
    if baggage_weight > 30:
        print("Status: Cleared for boarding (Extra baggage charge applies)")
    else:
        print("Status: Cleared for boarding")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
transaction_amount = float(input("Enter transaction amount: "))
location_match = input("Is the location matching? (yes/no): ").strip().lower()
time_of_transaction = input("Time of transaction (day/night): ").strip().lower()
previous_transactions_count = int(input("Enter previous transactions count in the last hour: "))
device_verified = input("Is the device verified? (yes/no): ").strip().lower()

if previous_transactions_count > 10:
    print("Status: Account Frozen (Too many transactions in 1 hour)")
elif location_match == "no" and transaction_amount > 100000:
    print("Status: Fraud Alert (Location mismatch + High amount)")
elif time_of_transaction == "night" and device_verified == "no":
    print("Status: High Risk (Night transaction + Unverified device)")
elif transaction_amount > 100000:
    print("Status: Suspicious Transaction (High amount)")
else:
    print("Status: Normal Transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal_record = input("Criminal record (yes/no): ").strip().lower()
number_of_cases = int(input("Enter number of cases: "))
travel_history_flagged = input("Is travel history flagged? (yes/no): ").strip().lower()
financial_activity_score = float(input("Enter financial activity score (0-100, where higher indicates suspicious activity): "))
last_activity_days = int(input("Days since last activity: "))
if (criminal_record == "yes" and number_of_cases > 3) or (travel_history_flagged == "yes" and financial_activity_score >= 70):
    print("Risk Level: High Risk")
elif last_activity_days > 365:
    print("Risk Level: Monitor")
elif criminal_record == "no" and number_of_cases == 0 and travel_history_flagged == "no" and financial_activity_score < 40:
    print("Risk Level: Low Risk")
else:
    print("Risk Level: Medium Risk")
```
